Task5:Dimensionality Reduction with PCA

**Loading the Inbuilt Dataset**

```python
from sklearn.datasets import load_wine
```

```python
wine = load_wine()
#which contains information about different varieties of wine.
#Each row represents a different wine sample, while each column represents a specific feature of that sample. 
```

**Dividing dataset into x and y**

```python
x = wine.data #these features represent quantitative measurements obtained from each sample.
y = wine.target #The target labels in the Wine dataset represent the class of each wine sample. 
```

```python
print(x)
```

```python
print(y)#There are three classes (0, 1, and 2), indicating different cultivars of wine.
```

```python
wine.data.shape
```

**Scaling the features**

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler = StandardScaler()
x_scaled = scaler.fit_transform(x)
```

**Implementing PCA**:
PCA is a dimensionality reduction technique that transforms the dataset into a new coordinate system.

```python
from sklearn.decomposition import PCA
```

```python
pca = PCA() #this instance will retain all principal components.
x_pca = pca.fit_transform(x_scaled)

#fit the PCA model to the scaled data and transformed it into the new coordinate system defined by the principal components.
#The transformed data, with reduced dimensionality,is stored in the variable x_pca.
```

```python
x_pca
```

**Finding Eigen Vectors and Eigen Values**

```python
eigenvectors = pca.components_
eigenvalues = pca.explained_variance_
#After fitting the PCA model,accessed the principal components (eigenvectors) and their corresponding eigenvalues.
#Eigenvectors represent the direction of maximum variance in the data, 
#while eigenvalues indicate the magnitude of variance along each eigenvector.
```

```python
eigenvectors
```

```python
eigenvalues
```

**Calculating Cumulative Explained Variance**

```python
explained_variance = pca.explained_variance_ratio_ #computed the explained variance ratio for each principal component.
```

```python
explained_variance
#The explained variance ratio indicates the proportion of variance explained by each principal component relative to 
#the total variance.
```

```python
import numpy as np
cumulative_explained_variance = np.cumsum(explained_variance)
#calculated the cumulative explained variance by taking the cumulative sum of the explained variance ratios.
```

```python
cumulative_explained_variance
```

**Visualizing Cumulative Explained Variance Ratio**

```python
import numpy as np
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))
plt.plot(range(1, x.shape[1] + 1),cumulative_explained_variance, marker='o', linestyle='-')
plt.xlabel('Number of Principal Components')
plt.ylabel('Cumulative Explained Variance Ratio')
plt.title('Cumulative Explained Variance Ratio by Principal Components')
plt.grid(True)
plt.show()
```

**Observations:**

1.The plot above illustrates the cumulative explained variance ratio by the number of principal components. Each point on the curve represents the cumulative proportion of total variance explained as we incrementally add principal components.

2.The curve will increase as the number of principal components increases. This is because each principal component captures some of the variance in the data.

3.The first principal component will capture the most variance, and subsequent components will capture progressively less variance. 

4.The more variance is explained by the first few principal components.

**Reducing Dimensions**

```python
chosen_components = int(np.argmax(cumulative_explained_variance >= 0.90))+1
#Based on the cumulative explained variance, 
#chose the number of principal components that explain at least 90% of the variance. adds 1 to get no.of components required.
```

```python
pca = PCA(n_components=chosen_components)
reduced_data = pca.fit_transform(x_scaled)
#instantiated PCA again with the chosen number of components and 
#transformed the scaled data to the reduced dimensionality space.
```

```python
print("Original wine data shape:", wine.data.shape)
print("Reduced wine data shape:", reduced_data.shape)
```
