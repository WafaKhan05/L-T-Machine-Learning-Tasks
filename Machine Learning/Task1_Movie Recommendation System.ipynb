```python
import pandas as pd
```

```python
movies_data = pd.read_csv("movies_metadata.csv", low_memory=False)
ratings = pd.read_csv("ratings.csv", low_memory=False)
```

```python
movies_df = pd.DataFrame(movies_data)
ratings_df = pd.DataFrame(ratings)
```

```python
movies_df
```

```python
ratings_df
```

```python
#since we had 2 different datasetsfor movies metadata and separate dataset for ratings.
#and both the datasets had data against the movie id.
#in ratings dataset movies were rated by different users. therefore, single movie has multiple ratings by different users.
#to get the actual mean of those ratings by different users we can group these ratings by movie ids and then get their mean.
rating_grouped_by_id=ratings_df.groupby('id')
rating_grouped_by_id
```

```python
ratings_mean=rating_grouped_by_id.mean()
ratings_mean
```

```python
#Delet the columns which are not required
ratings_mean.drop(columns='timestamp', inplace=True)
ratings_mean.drop(columns='userId', inplace=True)
```

```python
rating_mean_df = pd.DataFrame(ratings_mean)
rating_mean_df
```

```python
#Removing outliers data since column id contains some dates in approx 3 rows therefore removing the row entirely.
rows_to_delete = []
for idx, value in enumerate(movies_df['id']):
    try:
        int(value)  # Try to parse value to int
    except ValueError:  # If parsing fails, mark row for deletion
        rows_to_delete.append(idx)

# Drop rows marked for deletion
movies_df = movies_df.drop(rows_to_delete)
```

```python
#since id had some dates entries for some of the rows which are deleted in previous step, 
#initially python and pandas inferred its column type as object, therefore now setting datatype again back to int64.
movies_df['id'] = movies_df['id'].astype('int64')
```

```python
#now that we have ratings in another dataframe and movies metadata in another merging the both based on id.
#and with left join to ensure that we only get the ratings for which we have data in movies dataframe.
movies_w_rating = pd.merge(movies_df,rating_mean_df, on="id", how="left")
movies_w_rating
```

```python
movies_w_rating['rating'].isnull().value_counts()
```

```python
#for some of the movies we dont have ratings available, setting their values as mean of all the available ratings values.
import numpy as np
movies_w_rating["rating"].replace(np.nan,np.mean(movies_w_rating["rating"]),inplace=True)
movies_w_rating
```

```python
movies_w_rating.isnull().sum()
```

```python
#Delete the columns which are not required for recommendation 
movies_w_rating.drop(columns='homepage', inplace=True)
movies_w_rating.drop(columns='tagline', inplace=True)
movies_w_rating.drop(columns='overview', inplace=True)
movies_w_rating.drop(columns='poster_path', inplace=True)
movies_w_rating.drop(columns='imdb_id', inplace=True)
```

```python
#Replace null values with suitable values:
movies_w_rating["original_language"].replace(np.nan,"en",inplace=True)
movies_w_rating["popularity"].replace(np.nan,"not available",inplace=True)
movies_w_rating["production_companies"].replace(np.nan,"not availabe",inplace=True)
movies_w_rating["production_countries"].replace(np.nan,"USA",inplace=True)
movies_w_rating["release_date"].replace(np.nan,"20-08-1997",inplace=True)
movies_w_rating["revenue"].replace(np.nan,np.mean(movies_w_rating["revenue"]),inplace=True)
movies_w_rating["runtime"].replace(np.nan,np.mean(movies_w_rating["runtime"]),inplace=True)
movies_w_rating["spoken_languages"].replace(np.nan,"en",inplace=True)
movies_w_rating["status"].replace(np.nan,"Released",inplace=True)
movies_w_rating["title"].replace(np.nan,"not availabe",inplace=True)
movies_w_rating["video"].replace(np.nan,False,inplace=True)
movies_w_rating["vote_average"].replace(np.nan,np.mean(movies_w_rating["vote_average"]),inplace=True)
movies_w_rating["vote_count"].replace(np.nan,np.mean(movies_w_rating["vote_count"]),inplace=True)
movies_w_rating["belongs_to_collection"].replace(np.nan,"not availabe",inplace=True)
```

```python
#Now the data is cleaned
movies_w_rating.isnull().sum()
```

```python
movies_w_rating
```

```python
#Now we feature engineering that are relevent for making better recommendation
#Suppose user had watched movies whose id's are 8012, 11443, 11010
user_watched_movies_id = [8012, 11443, 11010]
#first we will find out avg of runtime and ratings
watched_movies_df = movies_w_rating[movies_w_rating['id'].isin(user_watched_movies_id)]
avg_runtime = watched_movies_df['runtime'].mean()
avg_rating = watched_movies_df['rating'].mean()
print('Average(Runtime):',avg_runtime)
print('Average(Rating):',avg_rating)
```

```python
#now we find the genre of the watched ids
watched_genres = []
for genres_list in watched_movies_df['genres']:
    print(genres_list)
```

```python

import json
watched_genres = []
for genres_list in watched_movies_df['genres']:
    g_list = json.loads(genres_list.replace("'", "\""))
    for genre in g_list:
        print(genre)
        watched_genres.append(genre['name'])
```

```python
watched_genres = list(set(watched_genres))
watched_genres
```

```python
# Convert each JSON string in the 'genres' column to a list of dictionaries
genres_list = movies_w_rating['genres'].apply(lambda x: json.loads(x.replace("'", "\"")))

# Check if any of the genres in the movie are in the list of watched genres
is_in_watched_genres = genres_list.apply(lambda x: any(genre['name'] in watched_genres for genre in x))

is_in_watched_genres
```

```python
# here we will set the criteria on the basis of which we recommend the movies:
# Criteria 1 : Check if the movies genre matches the user's fav genres i.e. his already watched movie genres.
# Criteria 2 : Check for the average runtime of the user's already watched movies +- 10 mins.
# Criteria 3 : Check if the rating of the movie is greater than or equal to user's already watched movies average rating.
# Criteria 4 : From all the available / matched movies, we will recommend the most popular movies to the user.
recommended_movies = movies_w_rating[
                            (is_in_watched_genres) & # Criteria 1
                            (movies_w_rating['runtime'] >= avg_runtime - 10) & # Criteria 2 -10 mins
                            (movies_w_rating['runtime'] <= avg_runtime + 10) & # Criteria 2 +10 mins
                            (movies_w_rating['rating'] >= avg_rating)] # Criteria 3 
recommended_movies
```

```python
recommended_movies = recommended_movies.sort_values(by='popularity', ascending=False) # Criteria 4
recommended_movies
```
